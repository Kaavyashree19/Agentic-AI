```python
%%writefile /content/distributed-puzzle-solver/puzzle.py

EXPLORERS = ["Arun", "Meera", "Kavin", "Divya", "Rahul"]

ISLANDS = [
    "Coral Island",
    "Emerald Island",
    "Golden Island",
    "Pearl Island",
    "Silver Island"
]

CLUES = [
    "Arun visited the Coral Island.",
    "The Falcon boat was used at the Coral Island.",
    "Meera discovered the Ruby.",
    "The Seagull boat was used at the Emerald Island.",
    "The Golden Crown was found on the Golden Island.",
    "Kavin used the Wave boat.",
    "Divya visited the Pearl Island.",
    "The Comet boat was used on the Pearl Island.",
    "Rahul discovered the Ancient Map.",
    "The Dolphin boat was used on the Silver Island.",
    "The Silver Coin was found on the Silver Island.",
    "The Emerald Gem was found on the Emerald Island.",
    "The Golden Crown was discovered by Kavin."
]

def solve_puzzle():

    explorer_island = {
        "Coral Island": "Arun",
        "Emerald Island": "Meera",
        "Golden Island": "Kavin",
        "Pearl Island": "Divya",
        "Silver Island": "Rahul"
    }

    boat_island = {
        "Coral Island": "Falcon",
        "Emerald Island": "Seagull",
        "Golden Island": "Wave",
        "Pearl Island": "Comet",
        "Silver Island": "Dolphin"
    }

    treasure_island = {
        "Coral Island": "Ruby",
        "Emerald Island": "Emerald Gem",
        "Golden Island": "Golden Crown",
        "Pearl Island": "Ancient Map",
        "Silver Island": "Silver Coin"
    }

    table = []

    for island in ISLANDS:
        table.append({
            "Island": island,
            "Explorer": explorer_island[island],
            "Boat": boat_island[island],
            "Treasure": treasure_island[island]
        })

    return {
        "explorer": "Kavin",
        "island": "Golden Island",
        "boat": "Wave",
        "treasure": "Golden Crown",
        "table": table
    }
```

```python
%%writefile /content/distributed-puzzle-solver/architectures.py

def centralized():

    return [
        ("Master Agent", "Received the puzzle."),
        ("Master Agent", "Dividing puzzle into subtasks."),
        ("Explorer Agent", "Analyzing explorer-island relationships."),
        ("Boat Agent", "Analyzing boat-island relationships."),
        ("Treasure Agent", "Analyzing treasure-island relationships."),
        ("Master Agent", "Collecting worker results."),
        ("Master Agent", "Checking consistency."),
        ("Master Agent", "Generating final solution.")
    ]


def hierarchical():

    return [
        ("Main Coordinator", "Received the puzzle."),
        ("Island Coordinator", "Managing island-related agents."),
        ("Explorer Agent", "Checking explorer-island relationships."),
        ("Position Agent", "Checking island constraints."),
        ("Resource Coordinator", "Managing resource agents."),
        ("Boat Agent", "Checking boat assignments."),
        ("Treasure Agent", "Checking treasure assignments."),
        ("Coordinators", "Sending validated results."),
        ("Main Coordinator", "Combining all results.")
    ]


def blackboard():

    return [
        ("Blackboard", "Puzzle posted to shared knowledge."),
        ("Explorer Agent", "Reading shared constraints."),
        ("Explorer Agent", "Posting explorer findings."),
        ("Boat Agent", "Reading updated knowledge."),
        ("Boat Agent", "Posting boat findings."),
        ("Treasure Agent", "Reading shared knowledge."),
        ("Treasure Agent", "Posting treasure findings."),
        ("Logic Agent", "Cross-validating knowledge."),
        ("Blackboard", "Knowledge converged."),
        ("Coordinator", "Retrieving final solution.")
    ]
```

```python
%%writefile /content/distributed-puzzle-solver/app.py

import streamlit as st
import time

from puzzle import CLUES, solve_puzzle
from architectures import centralized, hierarchical, blackboard


st.set_page_config(
    page_title="Multi-Agent Puzzle Solver",
    page_icon="🧩",
    layout="wide"
)


st.title("🧩 Multi-Agent Collaborative Puzzle Solver")

st.write(
    "Distributed Puzzle Solving using "
    "Centralized, Hierarchical and Blackboard Architectures"
)


st.sidebar.title("Architecture")

architecture = st.sidebar.selectbox(
    "Select Architecture",
    ["Centralized", "Hierarchical", "Blackboard"]
)


st.header("🏝️ Lost Treasure Island Puzzle")

st.write(
    "Five explorers visited five different islands. "
    "Each explorer used a different boat and discovered "
    "a different treasure."
)


st.subheader("📌 Puzzle Clues")

with st.expander("Show all clues"):

    for number, clue in enumerate(CLUES, 1):
        st.write(str(number) + ". " + clue)


st.subheader("🏗️ Selected Architecture")


if architecture == "Centralized":

    st.info(
        "A Master Agent controls worker agents "
        "and collects their results."
    )

    st.code(
        "Puzzle → Master Agent → Worker Agents → "
        "Master Agent → Final Answer"
    )


elif architecture == "Hierarchical":

    st.info(
        "A Main Coordinator manages specialized "
        "sub-coordinators and worker agents."
    )

    st.code(
        "Puzzle → Main Coordinator → Sub Coordinators → "
        "Workers → Final Answer"
    )


else:

    st.info(
        "Agents communicate through a shared Blackboard."
    )

    st.code(
        "Agents → Shared Blackboard → Logic Agent → "
        "Final Answer"
    )


st.divider()


if st.button(
    "🚀 START MULTI-AGENT SOLVING",
    type="primary",
    use_container_width=True
):

    solution = solve_puzzle()


    if architecture == "Centralized":
        logs = centralized()

    elif architecture == "Hierarchical":
        logs = hierarchical()

    else:
        logs = blackboard()


    st.subheader("🤖 Agent Collaboration")

    progress = st.progress(0)


    for number, data in enumerate(logs):

        agent = data[0]
        message = data[1]

        st.write(
            "**" + agent + "** : " + message
        )

        progress.progress(
            int((number + 1) / len(logs) * 100)
        )

        time.sleep(0.2)


    st.divider()

    st.subheader("🏆 Final Solution")


    col1, col2, col3, col4 = st.columns(4)


    col1.metric(
        "Explorer",
        solution["explorer"]
    )

    col2.metric(
        "Island",
        solution["island"]
    )

    col3.metric(
        "Boat",
        solution["boat"]
    )

    col4.metric(
        "Treasure",
        solution["treasure"]
    )


    st.success(
        "Kavin discovered the Golden Crown "
        "on the Golden Island using the Wave boat."
    )


    st.subheader("📊 Complete Solution")

    st.table(solution["table"])


    st.subheader("🔍 Architecture Result")


    if architecture == "Centralized":

        st.write(
            "The Master Agent coordinated the worker agents "
            "and integrated their results."
        )


    elif architecture == "Hierarchical":

        st.write(
            "The Main Coordinator collected results "
            "from specialized coordinators."
        )


    else:

        st.write(
            "The agents shared intermediate knowledge "
            "through the centralized Blackboard."
        )


st.divider()

st.caption(
    "Multi-Agent Systems Mini Project | "
    "Distributed Puzzle Solving"
)
```
